## 1. Introducción

En esta práctica vamos a crear un modelo de inteligencia artificial utilizando Google Teachable Machine.

El objetivo es entrenar un modelo capaz de reconocer dos objetos diferentes mediante imágenes: un teléfono móvil y una botella.

Después de entrenar el modelo, lo exportaremos a TensorFlow.js para utilizarlo en una página web creada con HTML, CSS y JavaScript.

Por último, añadiremos distintas mejoras a la aplicación, como iniciar y detener la cámara, establecer un porcentaje mínimo de confianza, guardar un historial de resultados y analizar imágenes subidas desde el ordenador.

## 2. Creación del modelo en Teachable Machine

Para comenzar la práctica, accedí a la página de [Google Teachable Machine](https://teachablemachine.withgoogle.com/) y creé un proyecto de tipo Image Project, seleccionando la opción Standard Image Model.

Una vez creado el proyecto, configuré dos categorías:

- **Móvil:** para reconocer teléfonos móviles.
- **Botella:** para reconocer botellas.

Para entrenar el modelo utilicé la webcam del ordenador, tomando 100 imágenes de cada categoría.

Durante la captura fui cambiando la posición y la inclinación de los objetos para que el modelo pudiera reconocerlos desde diferentes ángulos.

En total utilicé 200 imágenes de entrenamiento, repartidas entre las dos categorías.

## 3. Entrenamiento del modelo

Una vez recogidas las imágenes, pulsé el botón **Preparar modelo** y esperé a que terminara el entrenamiento.

En este proceso se utiliza aprendizaje supervisado, ya que el modelo aprende a partir de imágenes que previamente hemos clasificado como Móvil o Botella.

Cuando terminó el entrenamiento, utilicé la vista previa para comprobar su funcionamiento.

Al enseñar un móvil a la cámara, el modelo mostraba una probabilidad elevada para la categoría Móvil. Al mostrar una botella, sucedía lo mismo con su categoría.

Por tanto, las primeras pruebas fueron satisfactorias y el modelo consiguió distinguir ambos objetos.

## 4. Exportación del modelo a TensorFlow.js

Una vez comprobado que el modelo reconocía correctamente los dos objetos, procedí a exportarlo para poder utilizarlo en una página web.

Para ello, pulsé el botón Exportar modelo de Teachable Machine, seleccioné TensorFlow.js y descargué el modelo.

Al descomprimir el archivo ZIP obtuve tres archivos:

- **model.json:** contiene la estructura del modelo.
- **metadata.json:** contiene información sobre las categorías que hemos entrenado.
- **weights.bin:** contiene los parámetros aprendidos durante el entrenamiento.

Estos archivos son necesarios para utilizar el modelo desde JavaScript.

## 6. Funcionamiento de la aplicación

Una vez preparados los archivos, programé la aplicación utilizando HTML, CSS y JavaScript.

Para comprobar su funcionamiento, abrí el archivo `index.html` utilizando Live Server.

Al pulsar el botón Iniciar cámara, se activa la webcam y el modelo comienza a analizar las imágenes en tiempo real.

Debajo de la cámara aparecen los porcentajes de probabilidad de las dos categorías: Móvil y Botella.

El modelo compara lo que aparece en la cámara con los ejemplos utilizados durante el entrenamiento e intenta identificar el objeto.



## 7. Mejoras realizadas

Después de comprobar que el modelo funcionaba correctamente, realicé las siguientes mejoras:

**1. Mejorar el diseño de la página**

Utilicé CSS para cambiar el fondo, los colores, los botones y el aspecto general de la página, consiguiendo una presentación más clara.

**2. Hacer la página responsive**

Añadí una regla `@media` en CSS para adaptar los tamaños y espacios cuando se utiliza la aplicación desde una pantalla pequeña, como la de un móvil.

**3. Botones de inicio y parada**

Incorporé dos botones para iniciar y detener la cámara sin necesidad de cerrar la página.

**4. Mensaje personalizado de salida**

Añadí un mensaje que indica qué objeto ha reconocido la inteligencia artificial. Si no alcanza la confianza mínima, aparece un mensaje indicando que no se reconoce el objeto.

**5. Porcentaje mínimo de coincidencia**

Incorporé una barra deslizante que permite establecer una confianza mínima entre el 0 % y el 100 %.

Si la probabilidad del objeto es inferior al porcentaje establecido, la aplicación no acepta la detección.

**6. Historial de resultados**

Añadí una lista donde se guardan las últimas 10 detecciones realizadas.

También incorporé un botón para limpiar el historial. Los resultados se conservan mientras la página permanece abierta.

**7. Subir imágenes para clasificarlas**

Añadí una opción que permite seleccionar una imagen del ordenador para analizarla con el mismo modelo de inteligencia artificial.

La aplicación muestra el objeto identificado y su porcentaje de confianza, siempre que alcance el mínimo establecido.

## 8. Conclusiones

En esta práctica he aprendido a crear y entrenar un modelo de inteligencia artificial utilizando Google Teachable Machine.
